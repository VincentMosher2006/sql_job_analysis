```python
import pandas as pd
import matplotlib.pyplot as plt

## load CSV
df = pd.read_csv("C:/Users/vince/OneDrive - Kent State University/Desktop/sql_job_analysis/sql_project/2_top_paying_roles.csv")

# Count skills Frequency 
skills_counts = df["skills"].value_counts()

## plot
skills_counts.sort_values().plot(kind="barh", figsize=(8,6))
plt.xlabel("Frequency")
plt.ylabel("Skill")
plt.title("Top 10 Data Scientist Skills in 2023")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

## load CSV
df = pd.read_csv("C:/Users/vince/OneDrive - Kent State University/Desktop/sql_job_analysis/sql_project/1_top_paying_jobs.csv")
top_jobs=df
##plot
plt.figure(figsize=(10,6))
plt.barh(top_jobs["job_title"], top_jobs["salary_year_avg"], color="skyblue")
plt.ylabel("Position")
plt.title("Average Salary Distribution for Top 10 Paying Data Science Jobs in 2023")

plt.gca().invert_yaxis()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
